## Imports

```python
from langchain_ollama import ChatOllama
from langchain_ollama import OllamaEmbeddings
```

## Constants

```python
base_url="http://maia:11434/"
```

```python
# Initialize the Ollama LLM wrapper, specifying the model name
# Ensure the model (e.g., "llama3") has been pulled via the CLI
llm = ChatOllama(model="llama3.1", base_url=base_url)

# Invoke the model with a prompt
response = llm.invoke("Explain the benefits of local AI in simple terms.")

# Print the response

print(response.content)
```

```python
ollama_emb = OllamaEmbeddings(
    model="embeddinggemma",
    base_url=base_url
)
r1 = ollama_emb.embed_documents(
    [
        "Alpha is the first letter of Greek alphabet",
        "Beta is the second letter of Greek alphabet",
    ]
)
r2 = ollama_emb.embed_query(
    "What is the second letter of Greek alphabet"
)
```
